### Task -2
creating raw delta tables providing the manual schema and record ingestion time and source file name 

```sql
%python
#creating the delta table for customers_bronze
from pyspark.sql.types import *
from pyspark.sql.functions import *
cust_schema=StructType([
                StructField('customer_id',StringType(),True),
                StructField('name',StringType(),True),
                StructField('email',StringType(),True),
                StructField('dob',DateType(),True),
                StructField('signup_date',DateType(),True),
                StructField('location_id',StringType(),True),
])
df_cust=spark.read.schema(cust_schema).csv('/Volumes/kishoredb/cap_db/cap_files/dimensions/dim_customers.csv',header=True)
df_cust_f=df_cust.withColumn('ingestion_time',current_timestamp()).withColumn('source_file_name',df_cust["_metadata.file_path"])
df_cust_f.write.format('delta').saveAsTable('customers_bronze')
display(spark.sql('select * from customers_bronze limit 10'))
```

```sql
%python
# Creating the Delta table for the restaurants dataset using manual schema
res_schema=StructType([
                StructField('restaurant_id',StringType(),True),
                StructField('name',StringType(),True),
                StructField('cuisines',StringType(),True),
                StructField('location_id',StringType(),True),
                StructField('rating',DoubleType(),True),
                StructField('delivery_fee',DoubleType(),True),
])
df_res=spark.read.schema(res_schema).csv('/Volumes/kishoredb/cap_db/cap_files/dimensions/dim_restaurants.csv',header=True)
df_res_f=df_res.withColumn('ingestion_time',current_timestamp()).withColumn('source_file_name',df_res['_metadata.file_path'])
df_res_f.write.format('delta').saveAsTable('restaurants_bronze')
display(spark.sql('select * from restaurants_bronze limit 10'))
```

```sql
%python
# Creating delta table for locations dataset using Manual Schema
loc_schema=StructType([
                StructField('location_id',StringType(),True),
                StructField('area',StringType(),True),
                StructField('city',StringType(),True),
                StructField('state',StringType(),True),
                StructField('latitude',DoubleType(),True),
                StructField('longitude',DoubleType(),True),
])
df_loc=spark.read.schema(loc_schema).csv('/Volumes/kishoredb/cap_db/cap_files/dimensions/dim_locations.csv',header=True)
df_loc_f=df_loc.withColumn('ingestion_time', current_timestamp()).withColumn('source_file_name', df_loc['_metadata.file_path'])

df_loc_f.write.format('delta').saveAsTable('locations_bronze')
display(spark.sql('select * from locations_bronze limit 10'))
```

```sql
%python
# Creating delta table for delivery_agents dataset using Manual Schema
del_schema=StructType([
                StructField('agent_id',StringType(),True),
                StructField('name',StringType(),True),
                StructField('phone',StringType(),True),
                StructField('rating',DoubleType(),True)
])
df_del=spark.read.schema(del_schema).csv('/Volumes/kishoredb/cap_db/cap_files/dimensions/dim_delivery_agents.csv',header=True)
df_del_f=df_del.withColumn('ingestion_time', current_timestamp()).withColumn('source_file_name', df_del['_metadata.file_path'])
df_del_f.write.format('delta').saveAsTable('delivery_agents_bronze')
display(spark.sql('select * from delivery_agents_bronze limit 10'))
```

```sql
%python
# Creating delta table for menu_items dataset using Manual Schema
menu_schema=StructType([
                StructField('item_id',StringType(),True),
                StructField('restaurant_id',StringType(),True),
                StructField('item_name',StringType(),True),
                StructField('price',DoubleType(),True),
                StructField('category',StringType(),True)
])
menu_df=spark.read.schema(menu_schema).csv('/Volumes/kishoredb/cap_db/cap_files/dimensions/dim_menu_items.csv',header=True)
menu_df_f=menu_df.withColumn('ingestion_time', current_timestamp()).withColumn('source_file_name', menu_df['_metadata.file_path'])
menu_df_f.write.format('delta').saveAsTable('menu_items_bronze')
display(spark.sql('select * from menu_items_bronze limit 10'))
```

### Task -3
Create cleaned data delta tables 

```sql
-- Creating a cleaned delta table using customers_bronze table
create or replace table customers_filtered
as 
select * from customers_bronze 
where customer_id is not null and location_id is not null and signup_date is not null and dob is not null and email is not null and name is not null ;

-- Now I'm displaying is there any nulls are there or not there
select count(*) - count(1) as num_nulls from customers_filtered;
```

```sql
select * from customers_filtered limit 10;
```

```sql
-- Creating a cleaned delta table using restaurants_bronze table 
create or replace table restaurants_filtered
using delta
as 
select * from restaurants_bronze
where restaurant_id is not null and name is not null and cuisines is not null and location_id is not null and rating is not null and delivery_fee is not null ;

-- Displaying no of null values 
select count(*)-count(-1) as num_nulls from restaurants_filtered;
```

```sql
select * from restaurants_filtered limit 10;
```

```sql
-- Creating a cleaned delta table using the locations_bronze table
create or replace table locations_filtered 
as 
select * from locations_bronze 
where location_id is not null and area is not null and city is not null and state is not null and latitude is not null and longitude is not null ;

-- Displaying no of null values 
select count(*)-count(1) as num_nulls from locations_filtered;
```

```sql
select * from locations_filtered limit 10;
```

```sql
-- Creating a cleaned delta table using the delivery_agents_bronze table
create or replace table delivery_agents_filtered
as 
select * from delivery_agents_bronze 
where agent_id is not null and name is not null and phone is not null and rating is not null ;

-- Displaying no of null values 
select count(*)-count(1) as num_nulls from delivery_agents_filtered;
```

```sql
select * from delivery_agents_filtered limit 10;
```

```sql
-- Creating a cleaned delta table using the menu_items_bronze table
create or replace table menu_items_filtered
as 
select * from menu_items_bronze 
where item_id is not null and restaurant_id is not null and item_name is not null and price is not null and category is not null ;

-- Displaying no of null values 
select count(*)-count(1) as num_nulls from menu_items_filtered;
```

```sql
select * from menu_items_filtered limit 10;
```
